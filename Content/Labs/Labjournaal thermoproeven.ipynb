# **Procedure: Bepalen van de soortelijke warmte van water**

## **1. Voorbereiding**

1. Verzamel alle benodigde materialen: maatbeker, weegschaal, water, elektrisch verwarmingselement (10 Ω, 10 W), voedingsbron, thermometer of temperatuursensor en een stopwatch.
2. Controleer of alle apparatuur goed werkt: thermometer kalibreren indien nodig, stopwatch testen, elektrische voeding controleren.

---

## **2. Meten van de massa van het water**

1. Weeg de lege maatbeker.
2. Vul de maatbeker met een gewenste hoeveelheid water (bijv. 200–500 ml).
3. Weeg de maatbeker opnieuw en bereken de massa van het water:
   **m = m(gevulde beker) – m(lege beker)**.
   (Aangezien de dichtheid van water ≈ 1 g/ml, kan volume ≈ massa worden aangehouden als snelle controle.)

---

## **3. Opstellen van de meetopstelling**

1. Plaats het water in een hittebestendig bekerglas of laat het in de maatbeker indien die geschikt is voor verhitting.
2. Plaats het verwarmingselement in het water zodat het volledig onder is, maar zonder dat het de wanden raakt.
3. Plaats de thermometer of temperatuursensor in het water, zo dat de meetpunt volledig ondergedompeld is.
4. Zorg dat het water niet wordt geroerd door externe factoren; het roeren later handmatig of met een roerder mag, maar wees consistent.

---

## **4. Startcondities meten**

1. Meet en noteer de **begintemperatuur** van het water, ( T_0 ).
2. Zet de stopwatch klaar, maar start nog niet.

---

## **5. Water verwarmen**

1. Stel de voedingsbron zo in dat het verwarmingselement werkt binnen zijn vermogenslimieten (10 W max.).
2. Noteer het ingestelde **vermogen P** en/of de aangelegde **spanning U** en **stroom I** (zodat je kunt controleren: ( P = U * I )).
3. Start de voeding **én** de stopwatch tegelijk.
4. Verwarm het water gedurende een vooraf bepaalde tijd (bijv. 2–5 minuten).
5. Meet om de 30 seconden de temperatuur van het water en noteer deze in een tabel.
6. Roer af en toe zachtjes om temperatuurlagen te voorkomen.

---

## **6. Stoppen van de meting**

1. Schakel de voeding na de gekozen tijd uit.
2. Noteer de **eindtijd** en de **eindt temperatuut** ( T_e ).

---

## **7. Berekeningen**

1. Bereken de temperatuurstijging:
   **ΔT = T_e – T_0**
2. Bereken de toegepaste energie:
   **Q = P · t**
   waarbij
   – *P* = vermogen van het verwarmingselement (in watt)
   – *t* = tijd dat het element aan stond (in seconden)
3. Bereken de soortelijke warmte ( c ) van water met:
   [
   c = Q/(m*ΔT)
   ]
4. Vergelijk het resultaat met de bekende literatuurwaarde van water:
    c_water ≈ 4180 , {J/(kg·°C)} 

---

## **8. Afronding en evaluatie**

1. Noteer mogelijke foutbronnen (warmteverlies aan omgeving, meetonnauwkeurigheid, niet-uniforme menging).
2. Geef aan hoe de nauwkeurigheid van de proef verbeterd zou kunnen worden (isolatie, nauwkeuriger sensors, langer verwarmen, hogere bemonsteringsfrequentie).

```python
#Een grotere tube bevat meer gas → groter volume →
# de daadwerkelijke drukverandering in de spuit wordt verzwakt door het extra volume.
#Met een kleinere tube is het totale gasvolume ongeveer gelijk aan het spuitvolume, 
# waardoor de drukmetingen nauwkeuriger Boyle’s wet volgen.

U_1= 180 #bij 40 ml
U_2= 314 #bij 20 ml
U_3= 231 #bij 30 ml, stap van 10 ml
U_4= 525 #van 40 ml naar 10

#bij 40 ml hoort een druk van 1,022 bar


#nu met de langere tube, Ux , x staat voor aantal ml
U50= 160
U40= 180
U30= 226
U20= 300
U10= 460

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Lineair model
def model(U, a, b):
    return a * U + b

# Meetdata (korte tube)
U = np.array([180, 231, 314, 525])
P = np.array([
    1.022,
    1.022 * 40/30,
    2 * 1.022,
    4 * 1.022
])

# Curve fit uitvoeren
params, covariance = curve_fit(model, U, P)
a, b = params

print("a =", a)
print("b =", b)

# Fitted curve
U_fit = np.linspace(min(U), max(U), 200)
P_fit = model(U_fit, a, b)

# Plot
plt.scatter(U, P, color='blue', label='Meetpunten')
plt.plot(U_fit, P_fit, 'r', label='curvefit')
plt.xlabel("Sensor spanning U")
plt.ylabel("Druk (bar)")
plt.title("P= aU+b")
plt.legend()
plt.grid(True)
plt.show()

#Dus P= 0.009*U-0.688

```

```python
# Opgave 2: Vervang daarbij de kleine tube voor een langere en bepaal
#  het onbekende volume van de tube
#  met een volgende meetserie waarbij je de druk en het volume bepaald.

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# --- Meetdata (lange tube) ---
U = np.array([160, 180, 226, 300, 460])
V_s = np.array([50, 40, 30, 20, 10])  # spuitvolumes

# --- Druk volgens Boyle bij de korte tube (voor calibratie) ---
p40 = 1.022
P = p40 * 40 / V_s   # Alleen voor calibratie van U -> p

# --- Stap 1: Kalibratie U -> p ---
def linear(U, a, b):
    return a*U + b

(a, b), cov1 = curve_fit(linear, U, P)

print("Kalibratie:")
print("a =", a)
print("b =", b)

# Bepaal gemeten druk:
P_meas = linear(U, a, b)


# --- Stap 2: Fit om volume van tube te bepalen ---
# Boyle’s model met extra tubevolume
def boyle(Vs, Vt):
    return p40 * (40 + Vt) / (Vs + Vt)

Vt_opt, cov2 = curve_fit(boyle, V_s, P_meas)
Vt = Vt_opt[0]

print("\nGeschat tube volume V_t =", Vt, "mL")


# --- Plot ---
Vs_fine = np.linspace(10, 50, 200)
P_fit = boyle(Vs_fine, Vt)

plt.scatter(V_s, P_meas, label="Gemeten druk via sensor")
plt.plot(Vs_fine, P_fit, 'r', label=f"Boyle fit (V_t={Vt:.2f} mL)")
plt.xlabel("Spuitvolume (mL)")
plt.ylabel("Druk (bar)")
plt.title("P-V diagram bepaling tube volume")
plt.legend()
plt.grid()
plt.show()

```

```python
#buizenproef

#ex 1: straling, convectie, geleiding
#ex 2: de vergelijking klopt na subsitutie
#ex 3: delta_T is dus heel klein t.o.v. T_0, waardoor we de warmteoverdrachtscoefficient h als een constante kunnen beschouwen
#ex 4: de fout in het warmtetransport door straling die we maken voor het temperatuurbereik is best groot
#ex 5: buitenoppervlak= 0.858 m^2, warmtecapaciteit is 41.4 J/K van messing buis, T_omg= 293 K
#ex 6: ja, en met een dop is er alleen warmtestraling, zonder dop is er ook convectie
# Hier de data en de analyse

#1 data met dop eraf
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def exp_func(t, A, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (A * np.exp(-t/tau) + T_omg)  

buitenoppervlak = 0.858 # bepaal zelf in m^2
warmtecapaciteit = 41.4 # bepaal de warmtecapaciteit in J/K

times = np.array([0,20,40,60,80,100,120,140,160,180,200,220,240,260,280,300,320,340,360,380,400,420,440,460,480,500,520,540,560,580,600])
temps = np.array([51.5, 50.3, 49.2, 48.3,47.2, 45.8, 44.1,42.5,41.9,40.7,39.2,38.1,37.6,36.5,36.1,35.7,35.1,34.3,34.1,33.8,33.2,32.9,32.6,32.2,31.7,31.3,31.1,30.9,30.6,30.5,30.2]) + 273.15


# pas beginwaardes aan naar schatting
# Het aantal maxfev moet wellicht hoger voor goede convergentie van de waarde
popt, pocv = curve_fit(exp_func, times, temps, p0=[30, 500, 20], maxfev=5000)

A_exp, tau_exp, T_omg_exp = popt

y_fit = exp_func(times, *popt)

plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')

plt.plot(times, temps, 'bo', label='measurement')
plt.plot(times, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg_exp))

plt.legend()

plt.show()

h_exp = (warmtecapaciteit) / (tau_exp * buitenoppervlak)
 
print(h_exp) # warmteoverdrachtscoëfficiënt in kW/m^2 K
# Sla figuren op met  
# 
# plt.savefig("Figuren/naam.png", dpi=450)


```
